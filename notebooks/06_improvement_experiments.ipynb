# Phase 6A: Controlled Cost-Sensitive Learning

This notebook is a presentation layer over generated Phase 6A results. Core training and selection logic lives in `src.experiments.phase6a`. Final-test data is excluded.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RESULT_DIR = ROOT / 'results' / 'experiments' / 'phase6a'
results = json.loads((RESULT_DIR / 'phase6a_results.json').read_text())
candidates = pd.read_csv(RESULT_DIR / 'phase6a_candidates.csv')
validation = pd.read_csv(RESULT_DIR / 'phase6a_validation.csv')
```
